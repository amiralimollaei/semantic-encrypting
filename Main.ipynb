```python
from pathlib import Path
from secrets import token_bytes

from semantic_encrypting.codec import TextCodec
from semantic_encrypting.encryption import SemanticEncryptor

checkpoint = Path.home() / "Desktop/repos/hub/LiquidAI/LFM3-1.7B-Base"
encryptor = SemanticEncryptor(TextCodec(checkpoint, device="cuda"))
```

```python
text = "A notebook experiment in round-trip semantic encryption: turn plaintext into model-generated text, then recover the original plaintext exactly."
```

```python
context = "He sent a long comment on the blog post excitingly:\n\n"
```

```python
key = token_bytes(32)
```

```python
encrypted = encryptor.encrypt(text, key, context=context)
print(encrypted)
```

```python
restored = encryptor.decrypt(encrypted, key, context=context)
assert restored == text
print(restored)
```
